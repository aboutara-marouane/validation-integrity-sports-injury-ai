## Decision before confirmatory prospective modelling

The released weekly table can reproduce classification of the supplied injury-labelled records, but it does not currently establish a continuously observed prospective risk set. Notebook 07 must remain provisional until the source sampling rule and the meaning of the systematically absent 21 days are confirmed. The next confirmatory analysis should focus on direct weekly-method replication, non-overlapping 21-offset validation, and observation-process negative controls.
