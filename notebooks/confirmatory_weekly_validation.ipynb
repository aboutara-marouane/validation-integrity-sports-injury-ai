## Confirmatory conclusion

The original-style implementation reproduces the published weekly ROC-AUC, but this performance is sensitive to test-outcome-dependent normalization, overlapping fitting/calibration samples, and overlapping three-week prediction histories. Once safeguards are applied, discrimination is weaker and unstable. Once adjacent three-week histories are removed through all 21 prespecified offsets, every evaluated algorithm performs at chance and has approximately zero probability skill. The full-window model exceeds refitted label controls, indicating some structured association, but that association is not temporally robust and does not produce useful alert precision.

The defensible contribution is methodological: trustworthy validation changes the conclusion from apparent predictive success to evidence of observation-process and temporal-overlap sensitivity. These results do not support automated injury warnings, causal interpretations of feature importance, or prospective 28-day forecasting.
